# Funciones

```python
def generar_recomendaciones(nuevo_usuario, similares_df):
    """
    Genera recomendaciones basadas en la comparación del usuario
    con perfiles similares que tienen ingresos altos.
    """

    recomendaciones = []

    # Educación
    if similares_df['education.num'].min() > nuevo_usuario['education.num']:
        recomendaciones.append("Incrementar el nivel educativo")

    # Horas trabajadas
    if similares_df['hours.per.week'].mean() > nuevo_usuario['hours.per.week']:
        recomendaciones.append("Aumentar las horas trabajadas por semana")

    # Ocupación
    top_occupations = similares_df['occupation'].value_counts().head(3).index
    if nuevo_usuario['occupation'] not in top_occupations:
        recomendaciones.append(
            f"Evaluar un cambio hacia ocupaciones como: {', '.join(top_occupations)}"
        )

    return recomendaciones
```
